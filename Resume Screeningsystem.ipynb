```python
from abc import ABC , abstractmethod
#class and object
class Job:
    def __init__(self,title,required_skills,min_experience,max_salary):
        self.title = title
        self.required_skills = required_skills
        self.min_experience = min_experience
        self.max_salary = max_salary
        
#ABSTRACTION
class ScreeningEngine(ABC):
    @abstractmethod
    def screen(self,Job):
        pass

#ENCAPSULATION  
class Candidate(ScreeningEngine):
    def __init__(self,name,skills,expected_salary):
        self._name = name
        self._skills = skills
        self._expected_salary = expected_salary
    
    def screen(self,Job):
        matched_skills = []

        for skill in Job.required_skills:
            if skill.lower() in [s.lower() for s in self._skills]:
                matched_skills.append(skill)
        return matched_skills

    def check_salary(self,Job):
        if self._expected_salary <= Job.max_salary:
            return True
        else:
            return False

    def skill_percentage(self,Job):
        matched = self.screen(Job)
        total = len(Job.required_skills)

        percentage = (len(matched) / total) *100
        return percentage

    def final_screening(self,Job):
        skill_score = self.skill_percentage(Job)
        salary_status = self.check_salary(Job)
        experience_status = self.check_experience(Job)

        if skill_score >= 70 and salary_status and experience_status:
            return "Candidate Selected.."
        else:
            return "Candidate Rejected.."

    def check_experience(self,Job):
        return Job.min_experience == 0

#INHERITANCE
class FresherCandidate(Candidate):
    def __init__(self,name,skills,expected_salary,gpa):
        super().__init__(name,skills,expected_salary)
        self._gpa = gpa
        
fresher1 = FresherCandidate("Rakesh",["Python","SQL","NLP"],20000,6.0)
job1 = Job("Gen ai",["Python","SQL","NLP"],0,20000)
candidate1 = FresherCandidate(
    "Rakesh",
    ["Python", "SQL", "NLP"],
    20000,
    6.0
)


class ExperienceCandidate(Candidate):
    def __init__(self,name,skills,expected_salary,experience):
        super().__init__(name,skills,expected_salary)
        self._experience = experience

    def check_experience(self,job):
        if self._experience >= job.min_experience:
            return True
        else:
            return False

#Mulitiple Candidate storing
candidates = []
candidate1 = FresherCandidate("maddy",["Python","SQL","NLP"],20000,6.0)
candidate2 = ExperienceCandidate("Akash",["Python","NLP","RAG","GEN AI"],50000,4)
candidates.append(candidate1)
candidates.append(candidate2)

experience1 = ExperienceCandidate("Rakesh",["Python","SQL","NLP"],20000,4)
job2 = Job("Python Developer", ["Python", "SQL"], 2,20000)

print("\n---MULITIPLE CANDIDATE SCREENING---")


# HR RESUME SCREENING REPORT

print("=" * 45)
print("       HR RESUME SCREENING SYSTEM")
print("=" * 45)

print("\nJOB DETAILS")
print("-" * 45)
print("Job Title       :", job1.title)
print("Required Skills :", job1.required_skills)
print("Minimum Exp     :", job1.min_experience, "Years")
print("Maximum Salary  : ₹", job1.max_salary)

print("\nCANDIDATE SCREENING RESULTS")
print("=" * 45)

for candidate in candidates:

    print("\nCandidate Name  :", candidate._name)
    print("-" * 45)

    matched = candidate.screen(job1)
    skill_score = candidate.skill_percentage(job1)
    salary_status = candidate.check_salary(job1)
    experience_status = candidate.check_experience(job1)
    result = candidate.final_screening(job1)

    print("Matched Skills  :", matched)
    print("Skill Score     :", round(skill_score, 2), "%")
    print("Salary Status   :", "Within Budget" if salary_status else "Above Budget")
    print("Experience      :", "Eligible" if experience_status else "Not Eligible")
    print("Final Decision  :", result)

print("\n" + "=" * 45)
print("          SCREENING COMPLETED")
print("=" * 45)

        
```
