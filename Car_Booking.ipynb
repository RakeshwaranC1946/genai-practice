```python
customer_name = []
kms = []
costs = []

def create_booking():
    name = input("Enter Customer  Name :")
    km = int(input("Enter Kilometer :"))

    customer_name.append(name)
    kms.append(km)
    print("Booked Successfully..") 

def cost_calculation():
    costs.clear()

    for km in kms:
        cost = km * 15
        costs.append(cost)
        # print(costs)

    print("Cost Calculated successfully..")
# create_booking()
# cost_calculation()

def history ():
    print("\n-----RIDE HISTORY-----")

    for i in range(len(customer_name)):
        print("Customer :",customer_name[i])
        print("KM :",kms[i])
        print("Cost :",costs[i])
        print("----------------------")
# history()

def search_ride():
    search = input("Enter Customer name to search : ")

    found = False

    for i in range(len(customer_name)):
        if customer_name[i] == search:
            print("Customer found")
            print("KM :", kms[i])
            print("Cost :", costs[i])
            found = True

    if found == False:
        print("Customer Not Found..")
# search_ride()     
        
def revenue():
    total = 0
    for cost in costs:
        total = total + cost
        print("Total :",total)
# revenue()

while True:
    print("=====Car Booking Management=====")
    print("1.Create Booking")
    print("2.Cost Calculation")
    print("3.Ride History")
    print("4.Search Ride")
    print("5.Revenue")
    print("6.Exit")

    choice = int(input("Enter your choice"))

    if choice == 1:
       create_booking()
    elif choice == 2:
       cost_calculation()
    elif choice == 3:
       history()
    elif choice == 4:
       search_ride()
    elif choice == 5:
       revenue()
    elif choice == 6:
        print("Thank You..")
        break
    else:
        print("Invalid Choice..")
```
