```python
from datetime import datetime

class library:
    def __init__(self):
        self.book_name : ""
        self.book_id : ""
        self.issued_date = ""
        self.submit_date = ""
        self.period = 0
        
    def get_details(self):
        self.book_name = input("Enter Book Name :")
        self.book_id = input("Enter Book ID :")
        self.issued_date = input("Enter Issued Date (dd/mm/yyyy) :")
        self.submit_date = input("Enter Submit Date (dd/mm/yyyy) :")
        self.period = int(input("Enter Period of the Book :"))
        
    def calculate_fine(self):
        issued = datetime.strptime(self.issued_date, "%d/%m/%Y")
        submit = datetime.strptime(self.submit_date, "%d/%m/%Y")

        daysused = (submit - issued).days
        delayed_days = daysused - self.period

        if delayed_days > 5:
            fine = (delayed_days - 5) * 10
        else:
            fine = 0

        return fine

    def display(self):
        fine = self.calculate_fine()

        print("-----Library Management System-----")
        print("Book Name :",self.book_name)
        print("Book ID :",self.book_id)
        print("Issued Date :",self.issued_date)
        print("Submit Date :",self.submit_date)
        print("Period :",self.period)
        print("Fine :",fine)

obj = library()
obj.get_details()
obj.display()
```
